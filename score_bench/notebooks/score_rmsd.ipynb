```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
```

```python
natives_df = pd.read_csv("../../dev/45_synthetic_native_4/data/csvs/natives_5.csv", index_col=0)
natives_df.head()
```

```python
decoy_df = pd.read_csv("../data/266_decoys/rand1000_rand_w_score.csv", index_col=0)
decoy_df.head()
```

```python
fig, ax = plt.subplots(figsize=(10,10))
plt.scatter(decoy_df["rmsd"],decoy_df["xray_native_5_0"]+decoy_df["xray_native_5_1"], c="black", s=50)
plt.scatter([0],natives_df["xray_native_5_0"]+natives_df["xray_native_5_1"], marker="x", c="black", s=1000, linewidths=4)

ax.set_yticks([7.5, 9.5, 11.5])

ax.set_xticks([0, .5, 1])

plt.tick_params(axis='both', which='major', length=0, width=0, labelsize=40)  # Adjust the size as needed

for spine in ax.spines.values():
    spine.set_linewidth(4)  # or any desired thickness

plt.savefig(Path(Path.home(), "Desktop/score_rmsd.png"))

# plt.figure(figsize=(10,10))
# plt.scatter(decoy_df_major["rmsd_state_0"],decoy_df_major["xray_native_0"], c="black", s=10)
# plt.scatter(decoy_df_minor["rmsd_state_1"],decoy_df_minor["xray_native_0"], c="black", s=10)
# plt.scatter(natives_df["rmsd_0"],natives_df["xray_0"], marker="x", c="black", s=100)

# plt.ylim(1,11)
# plt.xlim(-.1,1.1)
# plt.tick_params(axis='both', which='major', labelsize=20)  # Adjust the size as needed

# plt.savefig(Path(Path.home(), "Desktop/score_rmsd.png"))
```
